```python
text = input("Enter a string: ")

vowels = 0
consonants = 0

for char in text.lower():
    if char.isalpha():
        if char in "aeiou":
            vowels += 1
        else:
            consonants += 1

words = len(text.split())
characters = len(text)

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Words:", words)
print("Characters:", characters)
```

```python
text = input("Enter a string: ")

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")

if str1 == str2:
    print("Both strings are equal")
elif str1 < str2:
    print(str1, "comes before", str2)
else:
    print(str1, "comes after", str2)
```

```python
text = "  Hello Python World  "

# strip()
clean_text = text.strip()
print("After strip:", clean_text)

# replace()
new_text = clean_text.replace("Python", "Java")
print("After replace:", new_text)

# split()
words = clean_text.split()
print("After split:", words)

# join()
joined = "-".join(words)
print("After join:", joined)
```

```python
a = 10
b = 5

print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a > 5 and b < 10)
print(a < 5 or b == 5)
print(not (a == b))
```

```python
# Numbers
print(bool(0))       # False
print(bool(10))      # True

# Strings
print(bool(""))      # False
print(bool("Hello")) # True

# Lists
print(bool([]))      # False
print(bool([1, 2]))  # True

# Dictionaries
print(bool({}))              # False
print(bool({"name": "John"})) # True
```

```python
def check():
    print("Function executed")
    return True


print("Using AND:")
result = False and check()
print("Result:", result)

print("\nUsing OR:")
result = True or check()
print("Result:", result)

print("\nAND with True:")
result = True and check()
print("Result:", result)

print("\nOR with False:")
result = False or check()
print("Result:", result)
```
